### Import Pandas

```python
import pandas as pd
```

# Creating a Series
### Series From List

```python
num=[10,20,30,40]
num_series=pd.Series(num)
print(num_series)
```

### Creating a Series from a List of Strings


```python
s=pd.Series(["Apple","Banana","Mango","Orange"])
print(s)
```

### Series with Custom Index

```python
mark=[85,90,75]
mark_series=pd.Series(mark,index=["Rahul","Anu","John"])
print(mark_series)
                      
```

### Creating Series From Dictionary

```python
student={
    "Rahul":85,
    "Anu":90,
    "John":75
}
student_series=pd.Series(student)
print(student_series)
```

### Sorting a Series - Ascending Order


```python
z=pd.Series([10,35,20,40,25,30])
print(z.sort_values())
```

### Sorting a Series - Descending Order


```python
z
```

```python
print(z.sort_values(ascending=False))
```

### Resetting the Index After Sorting


```python
s = pd.Series([25, 10, 45, 30, 15])
result=s.sort_values().reset_index(drop=True)
print(result)
```

# MISSING VALUES IN A SERIES

```python
import numpy as np
```

```python
s = pd.Series([10, 20, np.nan, 40, np.nan])
print(s)
```

### Checking for Missing Values - isna()


```python
print(s.isna())
```

### Removing Missing Values - dropna()


```python
print(s.dropna())
```

### Filling Missing Values - fillna()


```python
print(s.fillna(0))
```
