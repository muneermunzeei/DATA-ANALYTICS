#### IMPORTING THE DATASET

```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python
df=pd.read_csv("e_commerce.csv")
df.head()
```

#### CHECKING MISSING VALUES

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

#### HANDLING MISSING VALUES

```python
df["Product_Price"].fillna(df["Product_Price"].mean(),inplace=True)
```

```python
df["Discount_Rate"].fillna(df["Discount_Rate"].mean(),inplace=True)
```

```python
df["Product_Rating"].fillna(df["Product_Rating"].mean(),inplace=True)
```

```python
df["Number_of_Reviews"].fillna(df["Number_of_Reviews"].mean(),inplace=True)
```

```python
df["Stock_Availability"].fillna("Unknown",inplace=True)
```

```python
df["Days_to_Deliver"].fillna(df["Days_to_Deliver"].mean(),inplace=True)
```

```python
df["Return_Rate"].fillna(df["Return_Rate"].mean(),inplace=True)
```

```python
df["Category_ID"].fillna(df["Category_ID"].mode()[0],inplace=True)
```

#### RECHECK MISSING VALUES

```python
df.isna().sum()
```

#### CHECK DUPLICATE VALUES

```python
df.duplicated().sum()
```

#### CHECKING UNIQUE VALUES

```python
df.nunique()
```

```python
df["Stock_Availability"].unique()
```

```python
df["Stock_Availability"].value_counts()
```

#### FINAL VALIDATION

```python
print("Total Missing Values: ",df.isna().sum().sum())
print("Total Duplicate Values: ",df.duplicated().sum())
```

#### SAVING THE CLEANED DATASET

```python
df.to_csv("Cleaned_e_commerce.csv",index=False)
```
