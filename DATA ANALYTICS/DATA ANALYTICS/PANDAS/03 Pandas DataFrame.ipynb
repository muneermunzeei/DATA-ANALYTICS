# PANDAS DATAFRAME

## Creating a DataFrame From a Dictionary

```python
import pandas as pd
```

```python
data={
    "Name":["Anu","Rahul","Priya"],
    "Age":[21,22,20],
    "Mark":[85,90,88]
}
df=pd.DataFrame(data)
print(df)
```

## Creating a DataFrame from a List


```python
data=[
    ["Anu",21,85],
    ["Rahul",22,90],
    ["Priya",20,88]
]
df=pd.DataFrame(data,columns=["Name","Age","Mark"])
print(df)
```

## Creating a DataFrame with a Custom Index


```python
data={
    "Name":["Anu","Rahul","Priya"],
    "Mark":[85,90,88]
}
df=pd.DataFrame(data,index=["S1","S2","S3"])
print(df)
    
    
```

## Displaying Columns - df.columns


```python
print(df.columns)
```

## Selecting a Single Column


```python
data={
    "Name":["Anu","Rahul","Priya"],
    "Age":[21,22,20],
    "Mark":[85,90,88]
}
df=pd.DataFrame(data)
print(df)

```

```python
print(df["Name"])
```

## Selecting a Particular Row and Column

## Selecting Rows Using loc


```python
data={
    "Name":["Anu","Rahul","Priya"],
    "Mark":[85,90,88]
}
df=pd.DataFrame(data,index=["S1","S2","S3"])
print(df.loc["S3"])
    
    
```

## Adding a New Column


```python
df=pd.DataFrame({
    "Name":["Anu","Rahul","Priya"],
    "Mark":[85,90,88]
})
df["Grade"]=["A","A+","A"]
print(df)
```

## Modifying an Existing Column


```python
df["Mark"]=[85,95,90]
print(df)
```

## Adding a New Row


```python
df=pd.DataFrame({
    "Name":["Anu","Rahul","Priya"],
    "Mark":[85,90,88]
})
df.loc[3]=["Syam",88]
print(df)
```

## Removing a Column - drop()


```python
print(df.drop("Mark",axis=1))
```

##  Removing a Row - drop()


```python
print(df.drop(3))
```

## Basic Information - info()


```python
df=pd.DataFrame({
    "Name":["Anu","Rahul","Priya"],
    "Age":[21,22,20],
    "Mark":[85,90,88]
})
print(df.info())
```

## Basic Statistics - describe()


```python
df=pd.DataFrame({
    "Mark":[90,88,59,72,85]
})
print(df.describe())
```

## Filtering Rows


```python
df=pd.DataFrame({
    "Name":["Anu","Rahul","Priya"],
    "Mark":[85,90,75]
})
print(df[df["Mark"]>80])
```

## Sorting a DataFrame


```python
#sort mark in ascending order
df=pd.DataFrame({
    "Name":["Anu","Rahul","Priya","Arun"],
    "Mark":[85,90,88,75]
})
print(df.sort_values("Mark"))
```

```python
# sort mark in decsending order
print(df.sort_values("Mark",ascending=False))
```

## Checking for Missing Values - isna()


```python
import numpy as np
```

```python
df=pd.DataFrame({
    "Name":["Anu","Rahul","Priya"],
    "Mark":[85,np.nan,75]
})
print(df.isna())#isna() or isnull()
```

## Counting Missing Values


```python
print(df.isna().sum())
```

## Filling Missing Values - fillna()


```python
print(df.fillna(0))
```

## Removing Rows with Missing Values - dropna()


```python
df=df.dropna()
print(df)
```

## Viewing First and Last Rows - head() and tail()


```python
#print first rows
df=pd.DataFrame({
    "Name":["Anu","Rahul","Priya"],
    "Age":[21,22,20],
    "Mark":[85,90,88]
})
df=df.head() # print first 5 rows
print(df)
```

```python
# Print First Two Rows
df=df.head(2)
print(df)
```

```python
# Print Last Columns
df=df.tail() # print last five rows
```

```python
# Print Last Two rows
df=df.tail(2)
print(df)
```

## DataFrame Arithmetic


```python
df=pd.DataFrame({
    "Name":["Anu","Rahul","Priya"],
    "Age":[21,22,20],
    "Mark":[85,90,88]
})
df["Mark"] = df["Mark"] + 5 
print(df)

```

```python
df["Mark"] = df["Mark"] - 5 
print(df)
```
