# MERGING DATA USING PANDAS

## Basic Merging

```python
import pandas as pd
```

```python
df1=pd.DataFrame({
    "Id":[1,2,3],
    "Name":["Asha","Binu","Cathy"]
})
df2=pd.DataFrame({
    "Id":[1,2,4],
    "Marks":[85,90,75]
})
merged_df=pd.merge(df1,df2,on="Id")
print(merged_df)
    
```

# TYPES OF JOINS

## Inner Join(default)

### Only Common Rows:

```python
pd.merge(df1,df2,on="Id",how="inner")
```

## Left Join

### All Rows From Left DataFrame:

```python
pd.merge(df1,df2,on="Id",how="left")
```

## Right Join

### All Rows From Right DataFrame:

```python
pd.merge(df1,df2,on="Id",how="right")
```

## Outer Join

### All Rows From Both:

```python
pd.merge(df1,df2,on="Id",how="outer")
```

## Merge on Different Column Names

```python
df1=pd.DataFrame({
    "Id":[1,2,3],
    "Name":["Asha","Binu","Cathy"]
})
df2=pd.DataFrame({
    "Student_Id":[1,2,4],
    "Marks":[85,90,75]
})
merged_df=pd.merge(df1,df2,left_on="Id",right_on="Student_Id")
print(merged_df)
```

## Concatenation

#### Row Wise

```python
pd.concat([df1,df2],axis=0)
```

#### Column Wise

```python
pd.concat([df1,df2],axis=1)
```

## Apply

```python
mark=pd.Series([45,60,75,90])
def grade(x):
    if x>50:
        return "Pass"
    else:
        return "Fail"
print(mark.apply(grade))
```

## Apply Using Lambda Function

```python
mark=pd.Series([45,60,75,90])
result=mark.apply(lambda x:x+5)
print(result)
```

## Apply() On DataFrame Columns

```python
df=pd.DataFrame({
    "Math":[80,70,90],
    "Science":[85,60,95]
})
print(df.apply(max))
```
