## DATA INTEGRATION AND TRANSFORMATION TASK

```python
import pandas as pd
```

```python
df1=pd.DataFrame({
    "Roll_no":[1,2,3,4,5],
    "Name":["Binu","Sanju","Yamin","Anu","Sali"],
    "Mark":[70,75,89,95,90]
})
df2=pd.DataFrame({
    "Roll_no":[1,2,3,6,7],
    "Grade":["B","B+","A","A+","A+"],
    "Class":[8,9,10,11,12]
})
merged_df=pd.merge(df1,df2,on="Roll_no")
print(merged_df)
    
```

### Inner Join

```python
pd.merge(df1,df2,on="Roll_no",how="inner")
```

### Left Join

```python
pd.merge(df1,df2,on="Roll_no",how="left")
```

### Right Join

```python
pd.merge(df1,df2,on="Roll_no",how="right")
```

### Outer Join

```python
pd.merge(df1,df2,on="Roll_no",how="outer")
```

### Merge on Different Column Names

```python
df1=pd.DataFrame({
    "Roll_no":[1,2,3,4,5],
    "Name":["Binu","Sanju","Yamin","Anu","Sali"],
    "Mark":[70,75,89,95,90]
})
df2=pd.DataFrame({
    "ID":[1,2,3,6,7],
    "Grade":["B","B+","A","A+","A+"],
    "Class":[8,9,10,11,12]
})
merged_df=pd.merge(df1,df2,left_on="Roll_no",right_on="ID")
print(merged_df)
```

### Concatenation

#### Row wise

```python
pd.concat([df1,df2],axis=0)
```

#### Column wise

```python
pd.concat([df1,df2],axis=1)
```

### Apply

```python
mark=pd.Series([90,45,55,72,80])
def grade(x):
    if x>55:
        return "Pass"
    else:
        return "Fail"
print(mark.apply(grade))
```

### Apply using Lambda 

```python
mark=pd.Series([90,45,55,72,80])
result=mark.apply(lambda x:x+10)
print(result)
```

### Apply on DataFrame

```python
df=pd.DataFrame({
    "English":[72,45,80],
    "Malayalam":[79,35,70]
})
print(df.apply(min))
```

```python
df=pd.DataFrame({
    "English":[72,45,80],
    "Malayalam":[79,35,70]
})
print(df.apply(max))
```
