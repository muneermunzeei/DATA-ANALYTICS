## PANDAS SERIES TASK

```python
import pandas as pd
```

```python
# 1. Create a Pandas Series containing the values 10, 20, 30, 40, 50 .
x=pd.Series([ 10, 20, 30, 40, 50 ])
print(x)
```

```python
# 2. Create a Pandas Series from the list:["Apple", "Banana", "Mango", "Orange"] .
s=pd.Series(["Apple", "Banana", "Mango", "Orange"]) 
print(s)

```

```python
# 3. Create a Pandas Series with custom indexes A, B, C, D and values 100, 200,300,400
values=[100,200,300,400]
values_series=pd.Series(values,index=["A","B","C","D"])
print(values_series)
                        
```

```python
# 4. Create a Pandas Series and access the third element.
s=pd.Series([10,20,30,40,50])
print(s[2])
```

```python
# 5. Create a Pandas Series and display the first three elements.
s=pd.Series([10,20,30,40,50])
print(s[:3])
```

```python
# 6. Create a Pandas Series containing five numbers and find its sum and mean.
s=pd.Series([10,20,30,40,50])
print("SUM: ",s.sum())
print("MEAN:",s.mean())
```

```python
# 7. Create a Pandas Series containing student marks and find the maximum and minimum mark
mark=pd.Series([35,40,50,80,72])
print("Maximum Mark:",mark.max())
print("Minimum Mark:",mark.min())
```

```python
# 8. Create a Pandas Series containing numbers and display only the values greaterthan 50
s=pd.Series([25,30,55,60,21])
print(s[s>50])
```

```python
# 9. Create a Pandas Series containing duplicate values and display the unique values
s=pd.Series([10,20,30,10,40,30])
print(s.unique())
```

```python
# 10. Create a Pandas Series and display its index, values, data type, and number of elements
a=pd.Series([10,15,20,25,30,35])
print("INDEX: ",a.index)
print("VALUES: ",a.values)
print("DATA TYPE: ",a.dtype)
print("NO OF ELEMENTS: ",len(a))
```

```python
# 11. Create a Pandas Series from the following dictionary:{"A": 100, "B": 200, "C": 300, "D": 400}
dictionary={
    "A":100,
    "B":200,
    "C":300,
    "D":400
}
dictionary_series=pd.Series(dictionary)
print(dictionary_series)
```

```python
#12. Create a Pandas Series with indexes A, B, C and values 10, 20, 30 . Add a new element with index D and value 40 .
values=pd.Series([10,20,30],index=["A","B","C"])
values["D"]=40
print(values)

```

```python
# 13. Create a Pandas Series with indexes A, B, C, D and values 10, 20, 30, 40 .Remove the element with index C .
values=pd.Series([10,20,30,40],index=["A","B","C","D"])
print(values.drop("C"))
```

```python
# 14. Create a Pandas Series containing [25, 10, 45, 30, 15] and sort the values in ascending order
a=pd.Series([25, 10, 45, 30, 15])
print(a.sort_values())
```

```python
# 15. Create a Pandas Series containing [10, 20, 30, 40, 50] and check whether the value 30 exists in the Series.
s=pd.Series([10, 20, 30, 40, 50])
print(30 in s.values)
```
