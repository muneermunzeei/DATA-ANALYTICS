#### Major Category Distribution

```python
df["Major_Category"].value_counts()
```

**Observation:**  
STEM has the highest number of students (15,079), while Arts has the lowest number of students (5,933). Business, Humanities, and Medical fall between these two categories.

#### Primary Use Case Distribution

```python
df["Primary_Use_Case"].value_counts()
```

**Observation:**  
Debugging or Troubleshooting is the most common AI use case, while Direct Answer Generation is the least common.

#### AI Usage Across Years Of Study

```python
df.groupby("Year_of_Study")["Weekly_GenAI_Hours"].mean()
```

**Observation:**  
AI usage is fairly similar across all years of study, with Senior students showing the highest average usage.

#### AI Usage vs Traditional Study Hours

```python
# add new column(AI_Usage_Level)
df["AI_Usage_Level"]=pd.cut(
    df["Weekly_GenAI_Hours"],
    bins=[0,5,10,float("inf")],
    labels=["Low","Medium","High"]
)
```

```python
df[["Weekly_GenAI_Hours",
    "AI_Usage_Level"]].head()
```

```python
df.groupby("AI_Usage_Level")["Traditional_Study_Hours"].mean()
```

**Observation:**  
Low AI users have the highest average traditional study hours (11.80 hours), while high AI users have the lowest (10.19 hours). This shows a decreasing pattern in traditional study hours as AI usage increases.

#### Pre-Semester GPA vs Post-Semester GPA

```python
df[["Pre_Semester_GPA","Post_Semester_GPA"]].mean()
```

**Observation:**  
The average GPA increased from 3.14 before the semester to 3.34 after the semester.

#### Burnout Risk Level Distribution

```python
df["Burnout_Risk_Level"].value_counts()
```

**Observation:**  
Most students fall under the Medium burnout risk level (21,144), followed by Low (16,369) and High (12,487). This shows that the majority of students in the dataset have a medium level of burnout risk.

#### AI Usage and Burnout Risk

```python
pd.crosstab(df["AI_Usage_Level"],
            df["Burnout_Risk_Level"])
```

**Observation:**  
Low AI users mostly fall under low burnout risk, while high AI users have the highest number of students in the high burnout risk category. The dataset shows an increasing pattern between AI usage level and burnout risk level.

#### Paid AI Subscription Usage

```python
df["Paid_Subscription"].value_counts()
```

**Observation:**  
Most students do not have a paid AI subscription. Out of 50,000 students, 28,846 are non-subscribers, while 21,154 have a paid subscription.

#### Prompt Engineering Skill Distribution

```python
df["Prompt_Engineering_Skill"].value_counts()
```

**Observation:**  
Most students have a Beginner level of prompt engineering skill (18,495), followed by Intermediate (16,696) and Advanced (13,809). This shows that beginner-level prompt engineering skill is the most common among the students in the dataset.

#### AI Tool Diversity Distribution

```python
df["Tool_Diversity"].value_counts().sort_index()
```

**Observation:**  
Most students use 3 different AI tools (16,926), while the fewest students use 5 tools (5,022). The dataset shows that using 3 AI tools is the most common level of tool diversity among students.

#### Institutional AI Policy Distribution

```python
df["Institutional_Policy"].value_counts()
```

**Observation:**  
Most students are from institutions that allow AI use with citation. This shows that AI use is more commonly permitted than strictly banned in the dataset.

#### AI Usage Level vs Exam Anxiety

```python
df.groupby("AI_Usage_Level")["Anxiety_Level_During_Exams"].mean()
```


**Observation:**
AI usage level increases, the average exam anxiety score also increases. High AI users have the highest average anxiety score.

#### AI Usage Level vs Skill Retention Score

```python
df.groupby("AI_Usage_Level")["Skill_Retention_Score"].mean()
```

**Observation:**  
There is no clear pattern between AI usage level and skill retention score. Higher AI usage does not show a consistent increase or decrease in skill retention.
