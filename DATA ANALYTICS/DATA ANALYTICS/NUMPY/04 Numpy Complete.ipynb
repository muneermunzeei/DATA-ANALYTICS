### Array Slicing

```python
a=np.array([10,20,30,40,50])
print(a[1:4])
```

```python
a=np.array([[10,20,30],
            [40,50,60]])
print(a[:,1:])
```

### Data Types

```python
a=np.array([10,20,30],dtype=float)
print(a)
print(a.dtype)
```

### Array Reshaping

```python
a=np.array([10,20,30,40,50,60])
print(a.reshape(2,3))

      
```

```python
a=np.array([[1,2],
            [4,5]])
print(a.reshape(4))
           
```

### Array Iterating


```python
a=np.array([10,20,30])
for i in a:
    print(i)
```

```python
# Array iterating in multidiamensional array
a=np.array([[10,20],
           [30,40]])
for i in np.nditer(a):
    print(i)
```

### Joining Arrays

```python
a=np.array([1,2])
b=np.array([3,4])
print(np.concatenate((a,b)))
```

```python
# array joining with stack()
a=np.array([1,2])
b=np.array([3,4])
print(np.stack((a,b)))
```

```python
# array joining with vstack()
a=np.array([1,2])
b=np.array([3,4])
print(np.vstack((a,b)))
```

```python
# array joining with hstack()
a=np.array([1,2])
b=np.array([3,4])
print(np.hstack((a,b)))
```

### Spliting Arrays

```python
a=np.array([1,2,3,4])
print(np.array_split(a,2))
```

### Searching Arrays

```python
a=np.array([10,20,30,20,10])
print(np.where(a==20))
```

```python
a=np.array([5,8,10,15])
print(np.where(a>8))
```

### Sorting Arrays

```python
a=np.array([20,10,40,30,50])
print(np.sort(a))
```

```python
a=np.array(["Banana","Apple","cherry"])
print(np.sort(a))
```

### Filter Arrays

```python
a=np.array([10,20,25,30,35,40])
print(a[a>25])
```

```python
a=np.array([1,2,3,4,5,6])
print(a[a % 2==0])
```

### Array Mathematics

```python
a=np.array([10,20,30])
b=np.array([1,2,3,])
print(a+b)
print(a-b)
print(a*2)
print(a**2)
print(a//2)
```

### Axis Concept

```python
a=np.array([[1,2,3],
            [4,5,6]])
print(np.sum(a,axis=0))#column wise
print(np.sum(a,axis=1))#row wise
```
