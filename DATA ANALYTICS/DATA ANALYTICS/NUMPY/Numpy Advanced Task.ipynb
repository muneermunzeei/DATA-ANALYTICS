```python
import numpy as np
```

```python
# 1. Create a NumPy array containing 10, 20, 30, 40, 50 . Display the array, its shape, data type, size of each element, and total memory used.
a=np.array([10,20,30,40,50])
print(a)
print("Shape :",a.shape)
print("Data Type: ",a.dtype)
print("Size of each Elements :",a.itemsize)
print("Total Memory Used: ",a.nbytes)
```

```python
# 2. Create a 3 × 4 array containing the numbers from 1 to 12 . Display the second and third rows.
a=np.arange(1,13)
a=a.reshape(3,4)
print(a)
print(a[1:3,:])
```

```python
# 3. Create an array containing the numbers from 1 to 12 and convert it into a 3 × 4 array.
a=np.arange(1,13)
a=a.reshape(3,4)
print(a)

```

```python
# 4. Create two arrays containing [10, 20, 30] and [40, 50, 60] . Combine them so that the result contains two rows.
a=np.array([10,20,30])
b=np.array([40,50,60])
print(np.vstack((a,b)))
```

```python
# 5. Divide the array [10, 20, 30, 40, 50] into three smaller arrays.
a=np.array([10,20,30,40,50])
print(np.array_split(a,3))
```

```python
# 6. Given the array [5, 10, 15, 20, 25, 30] , display the positions of all elements greater than 15 .
a=np.array([5,10,15,20,25,30])
print(np.where(a>15))
```

```python
# 7. Given the array [12, 5, 20, 8, 30, 15] , display only the values between 10 and 25 .
a=np.array([12,5,20,8,30,15])
print(a[(a>10) & (a<25)])
```

```python
# 8. Given the array [4, 9, 16, 25] , calculate the square root of every element.
a=np.array([4,9,16,25])
print(np.sqrt(a))
```

```python
# 9. Given the array [1, 2, 3, 4] , calculate the exponential value and natural logarithm of every element.
a=np.array([1,2,3,4])
print(np.exp(a))
print(np.log(a))
```

```python
# 10. Given the array [10, 20, 30, 40, 50] , find the total, average, median,standard deviation, variance, smallest value, and largest value.
a=np.array([10,20,30,40,50])
print("Total: ",np.sum(a))
print("Average : ",np.mean(a))
print("Median : ",np.median(a))
print("Standard Deviation : ",np.std(a))
print("Variance : ",np.var(a))
print("Smallest Value: ",np.min(a))
print("Largest Value: ",np.max(a))
```

```python
# 11. Given the array [10, 20, 5, 40, 15] , find the positions of the smallest and largest elements.
a=np.array([10, 20, 5, 40, 15])
print("Position of Smallest Element: ",np.argmin(a))
print("Position of Largest Element: ",np.argmax(a))
```

```python
# 12 Find the sum of the values in each row and the sum of the values in each column.
a=np.array([[2, 4, 6],
            [8, 10, 12]])
print("Sum Of First Row :",np.sum(a[0,:]))
print("Sum Of Second Row :",np.sum(a[1,:]))
print("Sum Of First Column :",np.sum(a[:,0]))
print("Sum Of Second Column :",np.sum(a[:,1]))
print("Sum Of Third Column :",np.sum(a[:,2]))

```

```python
# 13. Create an array containing [10, 20, 30, 40] . Create a separate copy and a view of the array. Modify the arrays and observe the effect on the original array
a=np.array([10,20,30,40])
b=a.copy()
c=a.view()
b[0]=5
c[2]=25
print(a)
print(b)
print(c)
```

```python
# 14. Generate five random decimal numbers, five random integers between 1 and 20 , and five values from the list [10, 20, 30, 40] .
a=np.random.rand(5)
b=np.random.randint(1,20,5)
c=np.random.choice([10,20,30,40],5)
print(a)
print(b)
print(c)
```

```python
# 15 Print every element one by one, then convert the array into a one-dimensional array
a=np.array([[1,2,3],
            [4,5,6]])
for i in np.nditer(a):
    print(i)
print(a.flatten())
           
```
