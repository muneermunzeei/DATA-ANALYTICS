```python
#create a simple calculator using the users choice of operator(+,-,*,/)
num1=int(input("Enter a number:"))
num2=int(input("Enter a number:"))
operator=(input("Enter a operator(+,-,*,/):"))
if operator=="+":
  print(num1+num2)
elif operator=="-":
    print(num1-num2)
elif operator=="*":
    print(num1*num2)
elif operator=="/":
    print(num1/num2)
else:
    print("Invalid input")
```

```python
#find the greatest among three numbers using nested if
x=int(input("Enter first number:"))
y=int(input("Enter second number:"))
z=int(input("Enter third number:"))
if x>y and x>z:
    print("First number is greatest")
else:
    if y>x and y>z:
      print("Second number is greatest")
    else:
      print("Third number is greatest")
```

```python
#check whether three angles from a valid triangle
a=int(input("Enter the angle1:"))
b=int(input("Enter the angle2:"))
c=int(input("Enter the angle3:"))
if a+b+c==180:
    print("Valid triangle")
else:
    print("Not a valid triangle")
```

```python
#check whether three sides can form atriangle and identify whether it is equilateral,isoscales,or scalene
a=int(input("Enter the side1:"))
b=int(input("Enter the side2:"))
c=int(input("Enter the side3:"))
if a+b>c and a+c>b and b+c>a:
    if a==b==c:
        print("Equilateral triangle")
    elif a==b or b==c or c==a:
        print("Isoscales triangle")
    else:
        print("Scalene triangle")
else:
    print("Not a valid triangle")
```

```python
#calculate the electricity bill based on units consumed using slabs
unit=int(input("Enter the unit consumed:"))
if unit>0 and unit<=100:
    Total_Amount=(unit*2)
elif unit>100 and unit<=200:
    Total_Amount=(100*2)+(unit-100)*3
else:
    Total_Amount=(100*2)+(100*3)+(unit-200)*5
print("Total Amount:",Total_Amount)
         
```
