### PYTHON match STATEMENT

```python
#Display the full form of department code
code=str(input("Enter your department code(CS/IT/EC/ME):")).upper()
match code:
    case "CS":
        print("Computer Science")
    case "IT":
        print("Information Technology")
    case "EC":
        print("Electronics")
    case "ME":
        print("Mechanical")
    case _:
        print("invalid")
```

```python
#Display the month name based on a month number
month=int(input("Enter month number"))
match month:
    case 1:
        print("January")
    case 2:
        print("February")
    case 3:
        print("March")
    case 4:
        print("April")
    case 5:
        print("May")
    case 6:
        print("June")
    case 7:
        print("July")
    case 8:
        print("August")
    case 9:
        print("September")
    case 10:
        print("October")
    case 11:
        print("November")
    case 12:
        print("December")
    case _:
        print("Invalid")
```

```python
#create a simple calculator
num1=int(input("Enter a number:"))
num2=int(input("Enter a number:"))
operator=(input("Enter operator(+,-,*,/):"))
match operator:
    case "+":
        print(num1+num2)
    case "-":
        print(num1-num2)
    case "*":
        print(num1*num2)
    case "/":
        if num2!=0:
         print(num1/num2)
        else:
         print("divition not possible")
    case _:
        print("Invalid")
```

```python
# Display the grade description based on grade
grade=str(input("Enter your grade:")).upper()
match grade:
    case "A":
        print("Excellent")
    case "B":
        print("Good")
    case "C":
        print("Average")
    case "D":
        print("Pass")
    case "F":
        print("Fail")
    case _:
        print("Invalid")
```

```python
#Display the season based on a month
num=int(input("Enter month number:"))
match num:
    case 1|2|12:
        print("Winter")
    case 3|4|5:
        print("Summer")
    case 6|7|8|9:
        print("Monsoon")
    case 10|11:
        print("Spring")
    case _:
        print("Invalid")
    
```

```python
# create a menu driven program to select a drink
print("------------     MENU   -----------")
print("1. Tea")
print("2. Coffee")
print("3. Juice")
print("4. Water")

choice=int(input("Enter a choice(1-4):"))
match choice:
    case 1:
        print("you selected Tea")
    case 2:
        print("you selected Coffee")
    case 3:
         print("you selected Juice")
    case 4:
        print("you selected Water")
    case _:
        print("Invalid choice")




    
```

```python
#Display the number of days in a month using the month number
month=int(input("Enter month number:"))
match month:
    case 1|3|5|7|8|10|12:
        print("31 Days")
    case 4|6|9|11:
        print("30 Days")
    case 2:
        print("28 Days")
    case _:
        print("Invalid")
```
