### PYTHON SETS

```python
#create list with duplicates and remove duplicates
mylist=["car","bus","jeep","auto","bus","auto"]
myset=set(mylist)
mylist1=list(myset)
print(mylist1)
```

```python
mysetf={"abhi","rohan","jiny","razmi","john"}
mysetc={"roshan","jiny","shanu","abhi","shafi"}

#print all players
common=mysetf|mysetc
print("All Players: ",common)

#print the players who are present in both the football and cricket
mysetcf=mysetf&mysetc
print("Including Both :",mysetcf)

#print the players who are present in only one item
notcommon=mysetf^mysetc
print("Present Only One Item: ",notcommon)

#print only the football players
footballonly=mysetf-mysetc
print("Football Only: ",footballonly)

#print only the cricket players
cricketonly=mysetc-mysetf
print("Cricket Only: ",cricketonly)

```

```python
#find missing numbers from 1 to 20
myset={1,2,3,4,6,7,8,9,11,12,13,14,16,17,18,19,20}
myset2=range(1,21)
myset5=set(myset2)
myset3=myset5-myset
print("missing numbers:",myset3)
```

```python
#given two strings,print the characters that are common in both strings
x=set(input("enter a string: "))
y=set(input("enter next string: "))
print(x.intersection(y))
```
