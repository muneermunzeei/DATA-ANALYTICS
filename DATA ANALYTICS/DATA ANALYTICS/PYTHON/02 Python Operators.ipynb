### ARITHMETIC OPERATORS

```python
x=int(input("Enter a number:"))
y=int(input("Enter the next number:"))
print(x+y)
print(x-y)
print(x*y)
print(x/y)
print(x%y)
print(x**y)
print(x//y)
```
