```python
# Find the largest number in the sequence 12, 45, 7, 89, 34, 67 .
x=[12,45,7,89,34,67]
large=x[0]
for i in x:
    if i>large:
        large=i
print("Largest Number: ",large)
```

```python
#Find the smallest number in the sequence 56, 23, 89, 12, 45, 7 .
x=[56,23,89,12,45,7]
small=x[0]
for i in x:
    if i<small:
        small=i
print("Smallest Number:  ",small)
```

```python
# Count the number of uppercase and lowercase letters in "Python Data Science"
uppercount=0
lowercount=0
text="Python Data Science"
for char in text:
    if char.isupper():
        uppercount+=1
    elif char.islower():
        lowercount+=1
print("Total Number Of Uppercase Letters:  ",uppercount)
print("Total Number Of Lowercase Letters:  ",lowercount)

```

```python
# Check whether a given number is prime using a for loop.
num=int(input("Enter a Number:"))
count=0
for i in range(1,num+1):
    if num%i==0:
     count+=1
if count==2:
     print("Prime Number")
else:
     print("Not Prime Number")
```
