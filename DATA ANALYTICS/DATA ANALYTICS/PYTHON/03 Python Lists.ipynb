### PYTHON LISTS

```python
#create a list
mylist=["apple","banana","cherry"]
print(mylist)
#add new one
mylist.insert(2,"orange")
print(mylist)
#remove an item
mylist.remove("orange")
print(mylist)
#change first item
mylist[0]="kiwi"
print(mylist)
#remove a specified(second) item
mylist.pop(1)
print(mylist)
```

```python
list=[10,50,32,12,47,63,90]
x=(len(list))
print(x)
#find tatal sum
y=sum(list)
print(y)
#find highest value
h=max(list)
print("highest value:",h)
#find smallest value
s=min(list)
print("smallest value:",s)
#find average
avg=sum(list)/(len(list))
print("average:",avg)
```

```python
#sort list into ascending order
list=[10,50,32,12,47,63,90]
list.sort()
print("Ascending order:",list)
#sort list into descending order
list.sort(reverse=True)
print("Descending order:",list)
```

```python
#reverse the list
list=[10,50,32,12,47,63,90]
list.reverse()
print(list)
```
