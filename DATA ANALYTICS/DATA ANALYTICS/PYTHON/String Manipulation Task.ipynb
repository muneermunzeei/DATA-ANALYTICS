## Strings are Arrays

```python
# 1. Print the first character of the string "Python" .
text="Python"
print(text[0])
```

```python
# 2. Print the last character of the string "Programming" .
text="Programming"
print(text[-1])
```

```python
# 3. Print the third character of the string "Computer" .
text="Computer"
print(text[2])
```

```python
# 4. Display the length of the string "Data Science" .
text="Data Science"
print(len(text))
```

```python
# 5. Iterate through the string "Welcome" and print each character on a new line.
text="Welcome"
for i in text:
    print(i)
```

## String Slicing

```python
# 1. Print the first five characters of the string "Artificial Intelligence" .
text="Artificial Intelligence" 
print(text[0:5])
```

```python
# 2. Print the last four characters of the string "Technology" .
text="Technology"
print(text[-4:])
```

```python
# 3. Print the characters from index 2 to index 7 of the string "Programming" .
text="Programming"
print(text[2:8])
```

```python
# 4. Print every second character of the string "Python Programming" .
text="Python Programming"
print(text[::2])
```

```python
# 5. Reverse the string "Machine Learning" using slicing.
text="Machine Learning"
print(text[::-1])
```

## Modify Strings

```python
# 1. Convert the string "python programming" to uppercase.
text="python programming"
print(text.upper())
```

```python
# 2. Convert the string "DATA SCIENCE" to lowercase.
text="DATA SCIENCE"
print(text.lower())
```

```python
# 3. Remove the leading and trailing spaces from the string " Hello Python " .
text= " Hello Python " 
print(text.strip())
```

```python
# 4. Replace "Python" with "Java" in the string "I am learning Python" .
text="I am learning Python"
print(text.replace("Python","Java"))
```

```python
# 5. Split the string "Python is easy to learn" into a list of words.
text="Python is easy to learn"
print(text.split())
```

## String Concatenation


```python
# 1. Concatenate the strings "Hello" and "World" with a space.
x="Hello"
y="World"
print(x+" "+y)
```

```python
# 2. Concatenate the strings "Data" and "Science" .
x="Data"
y="Science"
print(x+y)
```

```python
# 3. Display a welcome message by concatenating "Welcome" and "Riya" .
x="Welcome"
y="Riya"
print(x+" "+y)
```

```python
# 4. Join the strings "Calicut" and "Kerala" using a comma.
x="Calicut"
y="Kerala"
print(x+","+y)
```

```python
# 5. Create an email address by concatenating "john" , "doe" , and "@gmail.com" .
first="john"
last="doe"
domain="@gmail.com"
print(first+last+domain)
```
