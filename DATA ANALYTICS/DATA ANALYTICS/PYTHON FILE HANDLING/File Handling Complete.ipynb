### Create and write to file

```python
with open("example.txt","w") as file:
    file.write("Welcome to the new file!")
```

### Read File

```python
with open("example.txt","r")as file:
    print(file.read())
```

### Overwrite File

```python
with open("example.txt","w")as file:
    file.write("Hello! This is Python")
```

### Append new line

```python
with open("example.txt","a")as file:
    file.write("\nStart Your Python")
```

### Read full content again

```python
with open("example.txt","r")as file:
    print(file.read())
```

### Read line by line

```python
with open("example.txt","r")as file:
    for x in file:
        print(x.strip())
```

### Read only one line

```python
with open("example.txt","r")as file:
    print(file.readline())
```

### Working With File Pointer

```python
with open("example.txt","r")as file:
    print(file.read(6))
    file.seek(0)
    print(file.read())
```

### Tell Current Position

```python
with open("example.txt","r")as file:
    print(file.tell())
    file.read(10)
    print(file.tell())
```

### Read lines as List

```python
with open("example.txt","r")as file:
    lines=file.readlines()
    print(lines)
```

### Writing Multiple Lines

```python
lines=["\nMufeeda\n","Fidha\n","Jalva\n"]
with open("example.txt","a")as file:
    file.writelines(lines)
```

### File Copy

```python
with open("example.txt","r")as src:
    content=src.read()
with open("copy.txt","w")as dest:
    dest.write(content)
    

    
```

### Rename File

```python
import os
os.rename("example.txt","new_example.txt")
```

### Delete File

```python
import os
os.remove("copy.txt")
```

### Working With CSV Files

```python
import csv
with open("data.csv","w",newline="")as file:
    text=csv.writer(file)
    text.writerow(["Name","Age"])
    text.writerow(["mufi","21"])

```

### Working With JSON Files

```python
import json
data={"Name":"mufeeda","Age":21}
with open("data.json","w")as file:
    json.dump(data,file)
    
with open("data.json","r")as file:
    print(json.load(file))

```
